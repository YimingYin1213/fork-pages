## 4. Lesson Part 1: What Is an Algorithm?

**Presenter: Luke Sanders · about 7 minutes**

An **algorithm** is a finite set of instructions that accomplish a specific task. A recipe is an algorithm. So is "add up tonight's donations and check whether we hit the goal."

### Why algorithms matter

Safe Passage Heals runs on decisions a computer can make quickly: *Did the Safe Nights campaign reach its goal? Does the self-defense workshop still have seats? Which supply is running lowest?* Each of those is an algorithm. If it's right, staff spend their time helping people. If it's subtly wrong, they make decisions from bad numbers.

### What code is used in an algorithm?

Every algorithm, in any language, is built from just **three building blocks**:

| Building block | What it does | Python | College Board pseudocode |
| --- | --- | --- | --- |
| **Sequencing** | Run steps in order, top to bottom | one line after another | one line after another |
| **Selection** | Choose what to do based on a condition | `if` / `else` | `IF` / `ELSE` |
| **Iteration** | Repeat steps | `for`, `while` | `FOR EACH`, `REPEAT UNTIL` |

### Reference Sheet

Keep this open during the hacks.

**Key terms**

| Term | Definition | Safe Passage Heals example |
| --- | --- | --- |
| **Algorithm** | A finite set of instructions that accomplish a specific task | Add up donations, then compare the total to the goal |
| **Equivalent algorithms** | Different code, same result **and** same side effects, for every input | Two ways to check whether a workshop has an open seat |
| **Result** | The value an algorithm returns or prints | `3` large gifts |
| **Side effect** | A change an algorithm makes besides its result | Sorting the donation log while looking for the largest gift |
| **Boundary input** | The input exactly at a limit, where `<` and `<=` disagree | `12` sign-ups when the workshop holds `12` |
| **Trace** | Stepping through code line by line to predict what it does | Writing down `count` after every loop pass |
| **Modify** | Change an existing algorithm to do a related job | Count the large gifts → collect the large gifts |
| **Combine** | Use two algorithms together, usually one's output feeds the other | Only look for the largest gift if there were any large gifts |

**Python vs. College Board pseudocode**

| | Python | College Board pseudocode |
| --- | --- | --- |
| Assign | `total = 0` | `total ← 0` |
| Output | `print(total)` | `DISPLAY(total)` |
| Loop over a list | `for amount in donations:` | `FOR EACH amount IN donations { }` |
| What's inside the loop or `if` | indentation | `{ }` |
| Procedure | `def total_of(amounts):` | `PROCEDURE totalOf(amounts) { }` |
| Add to a list | `big.append(amount)` | `APPEND(big, amount)` |
| Length of a list | `len(donations)` | `LENGTH(donations)` |
| First index | `0` | **`1`** |

**Patterns you'll reuse**

| Pattern | Shape |
| --- | --- |
| **Sum** | Start at 0, loop, add each item |
| **Count matches** | Start at 0, loop, add 1 when a condition is true |
| **Collect matches** | Start with `[]`, loop, append when a condition is true |
| **Find the max** | Assume the first item wins, loop, replace it when you find a bigger one |

**Common mistakes**

| Mistake | What it looks like | Fix |
| --- | --- | --- |
| Judging by looks | "These two loops are basically the same" | Test below, at, and above the limit |
| Line in the wrong block | A counter update one indent outside the `if` | Trace the loop and write the counter down after every pass |
| Ignoring side effects | Both return 250, but one reordered your list | Print your input list again after calling the algorithm |
| Pseudocode index 0 | `donations[0]` in pseudocode | Pseudocode lists start at `1` |

### Example A: All three building blocks in one algorithm

The **Safe Nights** campaign has a goal of $500. Here are tonight's donations, in the order they came in. Did we reach the goal?

````python
# CODE_RUNNER: 3.09 Example A - Did the Safe Nights campaign reach its goal?

# SEQUENCING: set up the data first, in order
donations = [25, 150, 40, 100, 10, 250]   # dollars, in the order received
goal = 500

# ITERATION: visit every donation and add it to a running total
total = 0
for amount in donations:
    total = total + amount

# SELECTION: decide which message to show
if total >= goal:
    print(f"Goal met! Safe Nights raised ${total}.")
else:
    print(f"Still need ${goal - total} to reach the goal.")
````

### Example B: Different code, same result (AAP-2.L.1)

The self-defense workshop holds **12 people**. Here are three different ways to ask "is there still a seat?" Instead of trusting one test, this runner tries the numbers **below, at, and above** the limit, which is how you actually check whether algorithms are equivalent.

````python
# CODE_RUNNER: 3.09 Example B - Three ways to check for an open seat

capacity = 12   # the self-defense workshop holds 12 people

# Version A: check "is there room?"
def has_seat_a(signed_up, capacity):
    if signed_up < capacity:
        return True
    else:
        return False

# Version B: flip the question to "is it full?" and swap the answers
def has_seat_b(signed_up, capacity):
    if signed_up >= capacity:
        return False
    else:
        return True

# Version C: no if statement at all. The comparison already IS True or False.
def has_seat_c(signed_up, capacity):
    return signed_up < capacity

# Test below, exactly at, and above the limit
for signed_up in [11, 12, 13]:
    a = has_seat_a(signed_up, capacity)
    b = has_seat_b(signed_up, capacity)
    c = has_seat_c(signed_up, capacity)
    print(f"{signed_up} signed up -> A: {a}, B: {b}, C: {c}, all agree: {a == b == c}")
````

All three agree on every test, so they're **equivalent**. Version C is the "conditional written as an equivalent Boolean expression" that College Board mentions.

**Try it:** change `<` to `<=` in version C only and run again. The `12` row is now the only row that disagrees. That's why you always test the boundary.

### Example C: Similar code, different result (AAP-2.L.2)

Staff want to know how many **large gifts ($100 or more)** came in tonight. Both algorithms below loop over the same list and count. They differ by **four spaces of indentation.**

**Don't run it yet.** Write down what you think each one prints.

````python
# CODE_RUNNER: 3.09 Example C - Predict both counts BEFORE you run

donations = [25, 150, 40, 100, 10, 250]

# Algorithm A: count gifts of $100 or more
count_a = 0
for amount in donations:
    if amount >= 100:
        count_a = count_a + 1       # inside the if: only counts large gifts
print("Algorithm A counted:", count_a)

# Algorithm B: look closely at the last line of the loop
count_b = 0
for amount in donations:
    if amount >= 100:
        count_b = count_b + 1
    count_b = count_b + 1           # outside the if: runs for EVERY donation
print("Algorithm B counted:", count_b)
````

**Trace it.** Algorithm B adds 1 for every large gift **and** 1 for every gift, period:

| Donation | Algorithm A `count_a` | Algorithm B `count_b` |
| --- | --- | --- |
| 25 | 0 | 0 + 1 = 1 |
| 150 | 1 | 2 + 1 = 3 |
| 40 | 1 | 3 + 1 = 4 |
| 100 | 2 | 5 + 1 = 6 |
| 10 | 2 | 6 + 1 = 7 |
| 250 | 3 | 8 + 1 = 9 |

A says **3**, which is right. B says **9**, which is "large gifts plus all gifts" and means nothing. No error, no crash, just a wrong report that someone might hand to the board.

### Example D: Same result, different side effect (AAP-2.L.2)

Staff send thank-you notes **in the order donations arrived**, so the order of the list matters. Both procedures below find the largest gift. Run it and look at the list **after** each call.

````python
# CODE_RUNNER: 3.09 Example D - Same answer, but one changes the list

def largest_a(amounts):
    # Find-the-max pattern: assume the first gift is biggest, then check the rest
    largest = amounts[0]
    for amount in amounts:
        if amount > largest:
            largest = amount
    return largest

def largest_b(amounts):
    amounts.sort()        # SIDE EFFECT: this reorders the caller's actual list
    return amounts[-1]    # the last item after sorting is the biggest

donation_log = [25, 150, 40, 100, 10, 250]
print("A returns:", largest_a(donation_log), "| log is now:", donation_log)
print("B returns:", largest_b(donation_log), "| log is now:", donation_log)
````

Both return **250**, so if you only check the answer they look equivalent. But `largest_b` **sorted the donation log**, so the thank-you notes would now go out in the wrong order. Two algorithms are only equivalent if their results **and** their side effects match.

### Example E: Modify and combine (AAP-2.M)

Most new algorithms aren't written from scratch. They're a pattern you already trust, **modified** a little or **combined** with another one.

````python
# CODE_RUNNER: 3.09 Example E - Modify one pattern, then combine two

donations = [25, 150, 40, 100, 10, 250]
threshold = 100

# Pattern: COUNT matches
def count_large(amounts, threshold):
    count = 0
    for amount in amounts:
        if amount >= threshold:
            count = count + 1
    return count

# MODIFIED version: same loop, same if. Only what it keeps track of changed.
def collect_large(amounts, threshold):
    large = []                        # was: count = 0
    for amount in amounts:
        if amount >= threshold:
            large.append(amount)      # was: count = count + 1
    return large                      # was: return count

# COMBINED: the first algorithm's answer decides whether we run the second
how_many = count_large(donations, threshold)
if how_many > 0:
    print(f"{how_many} large gifts tonight: {collect_large(donations, threshold)}")
    print(f"Largest single gift: ${max(donations)}")
else:
    print("No large gifts tonight.")
````

| Count version | Collect version |
| --- | --- |
| `count = 0` | `large = []` |
| `count = count + 1` | `large.append(amount)` |
| `return count` | `return large` |

The loop and the condition didn't change at all. That's what **modifying** an algorithm means: keep the structure, change what it accumulates.

## Summary

- An **algorithm** is a finite set of instructions built from **sequencing, selection, and iteration**.
- Two algorithms are **equivalent** only if they give the same result **and** the same side effects for **every** input.
- Test **below, at, and above** the limit. That's where "equivalent" algorithms usually turn out not to be.
- **Trace before you run.** A line in the wrong place gives a wrong number with no error.
- Build new algorithms by **modifying** patterns you trust and **combining** them.
